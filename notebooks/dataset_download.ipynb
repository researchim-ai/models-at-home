# Загрузка датасетов

Ноутбук для скачивания датасетов в `/app/datasets/`. В остальных ноутбуках при отсутствии файла вызывается `ensure_pretrain_dataset(...)` — датасет подгрузится автоматически.

```python
from pathlib import Path

paths = project_paths()
datasets_dir = paths['datasets']
datasets_dir.mkdir(parents=True, exist_ok=True)
print('Каталог датасетов:', datasets_dir)
print('Содержимое:', list(datasets_dir.iterdir()) if datasets_dir.exists() else [])
```

## Pretrain: FineWeb (если файла нет — скачается при первом обращении)

`ensure_pretrain_dataset(path)` проверяет наличие файла и при отсутствии скачивает с HuggingFace (streaming).

```python
# Стандартный путь для pretrain-ноутбуков
PRETRAIN_PATH = '/app/datasets/fineweb-2_train.jsonl'
path = ensure_pretrain_dataset(PRETRAIN_PATH, max_rows=5000)
print('Готово:', path)
```

## Ручная загрузка с другими параметрами

Можно указать репозиторий, лимит строк и путь сохранения.

```python
# FineWeb-Edu (меньше по размеру)
path_edu = ensure_pretrain_dataset(
    path='/app/datasets/fineweb-edu_train.jsonl',
    max_rows=20_000,
    repo_id='HuggingFaceFW/fineweb-edu',
    subset='default',
    split='train',
)
print('Saved:', path_edu)
```
