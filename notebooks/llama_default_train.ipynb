# LLaMA-style default training (Models at Home)

Этот ноутбук использует встроенный helper `train_default_llama(...)`, который автоматически доступен в Jupyter kernel startup.

```python
from pathlib import Path

paths = project_paths()
for k, v in paths.items():
    print(f"{k:>10}: {v}")
```

```python
# === Default config (same spirit as Studio LLaMA-style defaults) ===
DATA_PATH = "/app/datasets/fineweb-2_train.jsonl"  # поменяй на свой файл
ensure_pretrain_dataset(DATA_PATH)  # скачает с HF, если файла нет
OUTPUT_DIR = "/app/out/notebook_llama_default"
TOKENIZER = "gpt2"

HIDDEN_SIZE = 512
NUM_LAYERS = 8
N_HEADS = 8
SEQ_LEN = 2048

BATCH_SIZE = 8
GRAD_ACCUM = 4
EPOCHS = 1
LEARNING_RATE = 3e-4
WARMUP_STEPS = 100
```

```python
# Сначала покажем exact команду без запуска
train_default_llama(
    data_path=DATA_PATH,
    output_dir=OUTPUT_DIR,
    tokenizer_path=TOKENIZER,
    hidden_size=HIDDEN_SIZE,
    num_layers=NUM_LAYERS,
    n_heads=N_HEADS,
    seq_len=SEQ_LEN,
    batch_size=BATCH_SIZE,
    gradient_accumulation=GRAD_ACCUM,
    epochs=EPOCHS,
    learning_rate=LEARNING_RATE,
    warmup_steps=WARMUP_STEPS,
    dry_run=True,
)
```

```python
# Реальный запуск обучения
return_code = train_default_llama(
    data_path=DATA_PATH,
    output_dir=OUTPUT_DIR,
    tokenizer_path=TOKENIZER,
    hidden_size=HIDDEN_SIZE,
    num_layers=NUM_LAYERS,
    n_heads=N_HEADS,
    seq_len=SEQ_LEN,
    batch_size=BATCH_SIZE,
    gradient_accumulation=GRAD_ACCUM,
    epochs=EPOCHS,
    learning_rate=LEARNING_RATE,
    warmup_steps=WARMUP_STEPS,
)
print("training return_code:", return_code)
```
