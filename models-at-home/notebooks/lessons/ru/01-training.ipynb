# Как модель учится

Loss, градиенты, шаг оптимизатора и проверка на новых данных.

> Все примеры этого урока работают на CPU, не скачивают модели и используют искусственные данные. Они объясняют механизм; качество настоящего обучения проверяется отдельно.

## Что важно понять

Модель — функция с обучаемыми параметрами. Loss измеряет ошибку; backward вычисляет производные, а оптимизатор меняет параметры. Здесь мы восстанавливаем линейную зависимость из шумных наблюдений.

Делим данные до обучения. Validation не участвует в backward: она помогает заметить переобучение. Низкая ошибка на искусственных данных подтверждает работу примера, но не качество на вашей задаче.

```python
%matplotlib inline
import torch
from torch import nn
from torch.nn import functional as F
import matplotlib.pyplot as plt
torch.set_num_threads(1)
torch.manual_seed(42)
plt.rcParams.update({'figure.figsize': (7, 3), 'axes.spines.top': False, 'axes.spines.right': False})
print('PyTorch:', torch.__version__, '| device: CPU')
```

## Запустите эксперимент

Запускайте ячейки по порядку. Код можно менять; Shift + Enter выполняет текущую ячейку.

```python
x = torch.linspace(-2, 2, 120).unsqueeze(1)
y = 2.5 * x - 0.7 + 0.15 * torch.randn_like(x)
order = torch.randperm(len(x)); train, val = order[:90], order[90:]
model = nn.Linear(1, 1)
optimizer = torch.optim.AdamW(model.parameters(), lr=0.08)
history = []
for step in range(100):
    optimizer.zero_grad(set_to_none=True)
    loss = F.mse_loss(model(x[train]), y[train])
    loss.backward(); optimizer.step()
    history.append(loss.item())
with torch.no_grad():
    print('validation MSE:', F.mse_loss(model(x[val]), y[val]).item())
print('weight, bias:', model.weight.item(), model.bias.item())
plt.plot(history, color='#d73535'); plt.xlabel('step'); plt.ylabel('train loss'); plt.show()
```
